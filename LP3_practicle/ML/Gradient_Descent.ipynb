# Gradient Descent Algorithm to Find Local Minimum

**Aim:** Implement the Gradient Descent algorithm to find the local minimum of the function:

\[
y = (x+3)^2
\]

Starting point: `x = 2`

The minimum of this function is at `x = -3`, where `y = 0`. The algorithm iteratively updates the value of `x` using the derivative of the function.

## 1. Import required libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Define the function and its derivative

```python
# Function: y = (x + 3)^2
def function(x):
    return (x + 3) ** 2

# Derivative: dy/dx = 2(x + 3)
def derivative(x):
    return 2 * (x + 3)
```

## 3. Initialize parameters

```python
x = 2                 # Starting point
learning_rate = 0.1   # Step size
iterations = 50       # Number of iterations

x_history = []
y_history = []
```

## 4. Implement the Gradient Descent algorithm

```python
for i in range(iterations):
    x_history.append(x)
    y_history.append(function(x))

    # Update x in the opposite direction of the gradient
    x = x - learning_rate * derivative(x)

    print(
        f"Iteration {i + 1}: "
        f"x = {x:.6f}, y = {function(x):.6f}"
    )
```

## 5. Display the local minimum

```python
print("\\nLocal minimum found:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))
```

## 6. Plot the function and gradient descent path

```python
x_values = np.linspace(-5, 3, 200)
y_values = function(x_values)

plt.figure(figsize=(8, 5))
plt.plot(x_values, y_values, label='y = (x + 3)^2')
plt.scatter(x_history, y_history, color='red', s=20, label='Gradient descent steps')
plt.scatter(x, function(x), color='green', s=80, label='Final minimum')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent to Find Local Minimum')
plt.legend()
plt.grid(True)
plt.show()
```

## 7. Result and conclusion

The Gradient Descent algorithm was implemented for the function \(y=(x+3)^2\), starting from \(x=2\).

At each iteration, the value of `x` was updated in the opposite direction of the derivative. The algorithm converges towards the minimum point:

- **Minimum x-value:** approximately `-3`
- **Minimum y-value:** approximately `0`

Thus, Gradient Descent successfully finds the local (and global) minimum of the given function.
